```python
a = 20
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulo (remainder):", a % b)
print("Exponent:", a ** 2)
```

```python
first = "Sand"
last = "Technologies"

company = first + " " + last
print("I'm interning at", company)

# f-string version (more modern)
role = "Account Manager"
print(f"My role is {role} at {company}.")
```
