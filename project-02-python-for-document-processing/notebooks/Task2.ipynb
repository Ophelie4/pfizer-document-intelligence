```python
# Exercise 1: List Operations
fruits = ["mango", "banana", "strawberry", "pineapple", "apple"]

# Add a new fruit
fruits.append("papaya")

# Remove a fruit
fruits.remove("banana")

# Sort alphabetically
fruits.sort()

# Print first and last
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])
print("Full sorted list:", fruits)
```

```python
# Exercise 2: Tuple Operations
my_tuple = (4, 7, 2, 7, 9)

# Access second and fourth elements
print("Second element:", my_tuple[1])
print("Fourth element:", my_tuple[3])

# Create a new tuple by adding an element (tuples are immutable, so we concatenate)
new_tuple = my_tuple + (15,)
print("New tuple:", new_tuple)

# Count occurrences of a specific number
print("Occurrences of 7:", my_tuple.count(7))
```

```python
# Exercise 3: Set Operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Union
print("Union:", set_a | set_b)

# Intersection
print("Intersection:", set_a & set_b)

# Difference (elements in A but not B)
print("Difference (A - B):", set_a - set_b)

# Symmetric difference (elements in either but not both)
print("Symmetric difference:", set_a ^ set_b)
```

```python
# Exercise 4: Dictionary Operations
book = {
    "title": "The Alchemist",
    "author": "Paulo Coelho",
    "year": 1988
}

# Add genre
book["genre"] = "Fiction"

# Update year
book["year"] = 1993

# Remove author
del book["author"]

# Print all keys and values
for key, value in book.items():
    print(f"{key}: {value}")
```

```python
# Exercise 5: Slicing and Indexing
numbers = list(range(1, 21))  # [1, 2, 3, ..., 20]

print("First five:", numbers[:5])
print("Last five:", numbers[-5:])
print("Every second element:", numbers[::2])
print("Reversed:", numbers[::-1])
```
