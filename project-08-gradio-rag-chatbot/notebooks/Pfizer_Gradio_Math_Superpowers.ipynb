# Math Superpowers — a tiny Gradio app

One number in, three results out: **square**, **cube** and **factorial**.

`launch(share=True)` prints a public `*.gradio.live` link that works on a phone.
That link stays alive only while this notebook is running (72 hours max), so keep the
Colab tab open while anyone is testing it.

```python
!pip -q install gradio
import gradio as gr
import math
print("gradio", gr.__version__)
```

## The function

`math.factorial()` only accepts a non-negative whole number, so anything else gets a plain
explanation instead of a crash. Square and cube still come back either way.

```python
def math_superpowers(number):
    """Return the square, cube and factorial of one number as a dictionary."""
    if number is None:
        return {"error": "Type a number first."}

    results = {
        "number": number,
        "square": number ** 2,
        "cube": number ** 3,
    }

    if number < 0:
        results["factorial"] = "not defined for negative numbers"
    elif number != int(number):
        results["factorial"] = "only whole numbers have a factorial"
    elif number > 170:
        results["factorial"] = "too large to display (try 170 or less)"
    else:
        results["factorial"] = math.factorial(int(number))

    return results


for n in [5, 0, 1, 12, -3, 2.5]:
    print(n, "->", math_superpowers(n))
```

## The interface

```python
demo = gr.Interface(
    fn=math_superpowers,
    inputs=gr.Number(label="Your number", value=5),
    outputs=gr.JSON(label="Square, cube and factorial"),
    title="Math Superpowers",
    description="Give it one number and it hands back the square, the cube and the factorial.",
    examples=[[5], [10], [0], [20]],
    flagging_mode="never" if gr.__version__ >= "5" else None,
)

demo.launch(share=True)
```

```python
print(demo.share_url, demo.local_url)
```

## Notes

- `gr.Number` keeps the input numeric, so the function never receives a string it has to parse.
- Returning a dictionary into `gr.JSON` shows all three answers at once and keeps each one
  labelled, which a plain list would not.
- Swapping factorial for square root is a two-line change: `math.sqrt(number)` instead of
  `math.factorial(int(number))`, and the guard becomes "negative numbers have no real square
  root" rather than the whole-number check.
