```python
# Exercise 1: Even or Odd
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(f"{number} is Even")
else:
    print(f"{number} is Odd")
```

```python
# Exercise 2: Guess the Number
import random

secret = random.randint(1, 100)
guess = None

while guess != secret:
    guess = int(input("Guess the number (1-100): "))
    if guess < secret:
        print("Too low! Try again.")
    elif guess > secret:
        print("Too high! Try again.")
    else:
        print(f"Correct! The number was {secret}.")
```

```python
# Exercise 3: Print Prime Numbers up to 50
print("Prime numbers up to 50:")

for num in range(2, 51):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

```python
# Exercise 4: Factorial using a while loop
number = int(input("Enter a number: "))

factorial = 1
i = 1

while i <= number:
    factorial *= i
    i += 1

print(f"The factorial of {number} is {factorial}")
```

```python
# Exercise 5: Multiplication Table
number = int(input("Enter a number to see its multiplication table: "))

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
# Exercise 6: Count Vowels using for loop + continue
text = input("Enter a string: ")

vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char not in vowels:
        continue  # skip non-vowels
    count += 1

print(f"Number of vowels in '{text}': {count}")
```
